## 6. Drawdown a lo largo del tiempo

El drawdown mide la pérdida desde el máximo histórico alcanzado hasta cada fecha:

$$
DD_t = \frac{W_t}{\max_{s \leq t} W_s} - 1
$$

Permite visualizar no solo cuánto cae un activo, sino también cuánto tiempo permanece por debajo de su máximo anterior.

````python
from matplotlib.ticker import PercentFormatter

drawdowns = analyzer.drawdown_series()

ax = drawdowns.plot(
    figsize=(14, 7),
)

ax.set_title("Drawdown histórico")
ax.set_xlabel("Fecha")
ax.set_ylabel("Drawdown")
ax.yaxis.set_major_formatter(
    PercentFormatter(1.0)
)
ax.axhline(
    y=0,
    linewidth=1,
)
ax.grid(alpha=0.3)
ax.legend(title="Activo")

plt.tight_layout()
plt.show()
````

````python
(
    analyzer.max_drawdown()
    .mul(100)
    .sort_values()
    .round(2)
    .rename("Max Drawdown (%)")
)
````

## 7. Volatilidad móvil

La volatilidad móvil permite observar cómo cambia el riesgo de cada activo a lo largo del tiempo.

Se utiliza una ventana de 63 sesiones, aproximadamente un trimestre bursátil:

$$
\sigma_{t,63}
=
\operatorname{Std}(R_{t-62},\dots,R_t)\sqrt{252}
$$

La anualización se realiza multiplicando la volatilidad de la ventana por $\sqrt{252}$.

````python
ROLLING_WINDOW = 63

rolling_vol = analyzer.rolling_volatility(
    window=ROLLING_WINDOW,
)

ax = rolling_vol.plot(
    figsize=(14, 7),
)

ax.set_title(
    f"Volatilidad anualizada móvil ({ROLLING_WINDOW} sesiones)"
)
ax.set_xlabel("Fecha")
ax.set_ylabel("Volatilidad anualizada")
ax.yaxis.set_major_formatter(
    PercentFormatter(1.0)
)
ax.grid(alpha=0.3)
ax.legend(title="Activo")

plt.tight_layout()
plt.show()
````

## 8. Distribución de retornos, VaR y CVaR

Para cada activo se analiza la distribución histórica de retornos diarios.

El Value at Risk histórico al 95% indica el umbral de pérdida asociado al 5% peor de los retornos:

$$
VaR_{95\%} = -Q_{5\%}(R)
$$

El Conditional Value at Risk, o Expected Shortfall, mide la pérdida media una vez que nos encontramos dentro de esa cola:

$$
CVaR_{95\%}
=
-\mathbb{E}
\left[
R \mid R \leq Q_{5\%}(R)
\right]
$$

En los gráficos, como VaR y CVaR están almacenados como magnitudes de pérdida positivas, se representan sobre el eje de retornos como valores negativos.

````python
var_95 = analyzer.var(
    confidence_level=0.95,
    method="historical",
)

cvar_95 = analyzer.cvar(
    confidence_level=0.95,
    method="historical",
)

for ticker in returns.columns:
    fig, ax = plt.subplots(figsize=(11, 5))

    ax.hist(
        returns[ticker].dropna(),
        bins=60,
        alpha=0.75,
        edgecolor="black",
    )

    ax.axvline(
        -var_95[ticker],
        linestyle="--",
        linewidth=2,
        label=f"VaR 95% = {var_95[ticker]:.2%}",
    )

    ax.axvline(
        -cvar_95[ticker],
        linestyle=":",
        linewidth=2,
        label=f"CVaR 95% = {cvar_95[ticker]:.2%}",
    )

    ax.set_title(
        f"{ticker} — Distribución de retornos diarios"
    )
    ax.set_xlabel("Retorno diario")
    ax.set_ylabel("Frecuencia")
    ax.xaxis.set_major_formatter(
        PercentFormatter(1.0)
    )

    ax.grid(alpha=0.25)
    ax.legend()

    plt.tight_layout()
    plt.show()
````

````python
tail_risk = pd.DataFrame({
    "VaR Historical 95%": var_95,
    "CVaR Historical 95%": cvar_95,
    "Skewness": analyzer.skewness(),
    "Kurtosis": analyzer.kurtosis(),
})

tail_risk[
    [
        "VaR Historical 95%",
        "CVaR Historical 95%",
    ]
] *= 100

tail_risk.round(3).sort_values(
    "CVaR Historical 95%",
    ascending=False,
)
````

## 9. Comparación entre Sharpe y Calmar

El Sharpe ratio mide la rentabilidad excedente obtenida por unidad de volatilidad total:

$$
Sharpe =
\frac{E[R-R_f]}{\sigma(R-R_f)}
$$

El Calmar ratio utiliza una visión distinta del riesgo:

$$
Calmar =
\frac{CAGR}{|Max\ Drawdown|}
$$

Por tanto, dos activos pueden obtener rankings diferentes si uno presenta una volatilidad moderada pero sufre drawdowns profundos o persistentes.

````python
ratio_comparison = pd.DataFrame({
    "Sharpe": analyzer.sharpe(),
    "Calmar": analyzer.calmar(),
})

ratio_comparison
````

````python
ratio_comparison["Sharpe Rank"] = (
    ratio_comparison["Sharpe"]
    .rank(ascending=False, method="min")
    .astype(int)
)

ratio_comparison["Calmar Rank"] = (
    ratio_comparison["Calmar"]
    .rank(ascending=False, method="min")
    .astype(int)
)

ratio_comparison = ratio_comparison.sort_values(
    "Sharpe Rank"
)

ratio_comparison.round(3)
````

````python
same_ranking = (
    ratio_comparison["Sharpe Rank"]
    == ratio_comparison["Calmar Rank"]
).all()

print(
    "¿Coinciden completamente los rankings?",
    same_ranking,
)
````

````python
ratio_comparison["Rank Difference"] = (
    ratio_comparison["Sharpe Rank"]
    - ratio_comparison["Calmar Rank"]
)

ratio_comparison.round(3)
````

````python
ax = ratio_comparison[
    ["Sharpe", "Calmar"]
].plot(
    kind="bar",
    figsize=(11, 6),
)

ax.set_title(
    "Comparación de Sharpe y Calmar"
)
ax.set_xlabel("Activo")
ax.set_ylabel("Ratio")
ax.grid(
    axis="y",
    alpha=0.3,
)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````

## 10. VaR paramétrico en dos regímenes de mercado

Para estudiar la estabilidad del VaR paramétrico se comparan dos periodos:

- **2017–2019**: periodo previo a COVID.
- **2020–2022**: periodo marcado por la crisis COVID, la recuperación posterior, inflación y cambios importantes en los tipos de interés.

Se calculan dos versiones:

1. **VaR gaussiano**, que supone normalidad de los retornos.
2. **VaR Cornish-Fisher**, que ajusta el cuantil normal utilizando skewness y kurtosis.

Esto permite analizar tanto el cambio en el nivel de riesgo como el efecto de desviarse de la hipótesis de normalidad.

````python
period_1 = returns.loc[
    "2017-01-01":"2019-12-31"
]

period_2 = returns.loc[
    "2020-01-01":"2022-12-31"
]

print("Periodo 1:", period_1.index.min().date(), "→", period_1.index.max().date())
print("Observaciones:", len(period_1))

print()

print("Periodo 2:", period_2.index.min().date(), "→", period_2.index.max().date())
print("Observaciones:", len(period_2))
````

````python
analyzer_period_1 = RiskAnalyzer(
    period_1,
    periods_per_year=PERIODS_PER_YEAR,
)

analyzer_period_2 = RiskAnalyzer(
    period_2,
    periods_per_year=PERIODS_PER_YEAR,
)
````

````python
var_gaussian_comparison = pd.DataFrame({
    "2017-2019": analyzer_period_1.var(
        confidence_level=0.95,
        method="gaussian",
    ),
    "2020-2022": analyzer_period_2.var(
        confidence_level=0.95,
        method="gaussian",
    ),
})

var_gaussian_comparison *= 100

var_gaussian_comparison["Cambio (%)"] = (
    (
        var_gaussian_comparison["2020-2022"]
        / var_gaussian_comparison["2017-2019"]
    )
    - 1
) * 100

var_gaussian_comparison.round(2)
````

````python
var_cf_comparison = pd.DataFrame({
    "2017-2019": analyzer_period_1.var(
        confidence_level=0.95,
        method="cornish-fisher",
    ),
    "2020-2022": analyzer_period_2.var(
        confidence_level=0.95,
        method="cornish-fisher",
    ),
})

var_cf_comparison *= 100

var_cf_comparison.round(2)
````

````python
var_regimes = pd.DataFrame({
    "Gaussian 2017-2019": (
        analyzer_period_1.var(
            confidence_level=0.95,
            method="gaussian",
        ) * 100
    ),
    "Cornish-Fisher 2017-2019": (
        analyzer_period_1.var(
            confidence_level=0.95,
            method="cornish-fisher",
        ) * 100
    ),
    "Gaussian 2020-2022": (
        analyzer_period_2.var(
            confidence_level=0.95,
            method="gaussian",
        ) * 100
    ),
    "Cornish-Fisher 2020-2022": (
        analyzer_period_2.var(
            confidence_level=0.95,
            method="cornish-fisher",
        ) * 100
    ),
})

var_regimes.round(2)
````

````python
ax = var_regimes.plot(
    kind="bar",
    figsize=(13, 7),
)

ax.set_title(
    "VaR paramétrico al 95% en dos regímenes de mercado"
)
ax.set_xlabel("Activo")
ax.set_ylabel("VaR diario (%)")
ax.grid(
    axis="y",
    alpha=0.3,
)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````

````python
distribution_regimes = pd.DataFrame({
    "Skew 2017-2019": analyzer_period_1.skewness(),
    "Kurtosis 2017-2019": analyzer_period_1.kurtosis(),
    "Skew 2020-2022": analyzer_period_2.skewness(),
    "Kurtosis 2020-2022": analyzer_period_2.kurtosis(),
})

distribution_regimes.round(3)
````

## 11. Capital Resilience Ratio

Se propone una métrica propia denominada **Capital Resilience Ratio (CRR)**.

Su objetivo es medir cuánto crecimiento compuesto obtiene una inversión en relación con tres dimensiones distintas del riesgo:

1. profundidad de las pérdidas, mediante el máximo drawdown,
2. severidad de la cola, mediante el CVaR,
3. persistencia de las pérdidas, mediante el máximo tiempo bajo agua.

La métrica se define como:

$$
CRR =
\frac{CAGR}
{
\sqrt{|MDD|\cdot CVaR_{95}}
\left(
1+\frac{TUW_{max}}{N}
\right)
}
$$

donde:

- $MDD$ es el máximo drawdown,
- $CVaR_{95}$ es el Expected Shortfall histórico al 95%,
- $TUW_{max}$ es el máximo número consecutivo de periodos por debajo del máximo anterior,
- $N$ es el número de periodos por año.

Un CRR mayor indica una mayor capacidad de generar crecimiento compuesto frente a la profundidad, severidad y duración de las pérdidas.

La métrica se utiliza principalmente de forma **relativa para comparar carteras**, no mediante umbrales absolutos prefijados.

````python
crr = analyzer.capital_resilience(
    confidence_level=0.95,
)

crr_table = (
    crr
    .sort_values(ascending=False)
    .rename("Capital Resilience Ratio")
    .to_frame()
)

crr_table.round(3)
````

````python
crr_components = pd.DataFrame({
    "CAGR (%)": analyzer.cagr() * 100,
    "Max Drawdown (%)": analyzer.max_drawdown() * 100,
    "CVaR 95% (%)": analyzer.cvar(
        confidence_level=0.95,
        method="historical",
    ) * 100,
    "Max Time Under Water": analyzer.time_under_water().max(),
    "CRR": crr,
})

crr_components = crr_components.sort_values(
    "CRR",
    ascending=False,
)

crr_components.round(2)
````

````python
ax = crr.sort_values().plot(
    kind="barh",
    figsize=(10, 5),
)

ax.set_title(
    "Capital Resilience Ratio"
)
ax.set_xlabel("CRR")
ax.set_ylabel("Activo")
ax.grid(
    axis="x",
    alpha=0.3,
)

plt.tight_layout()
plt.show()
````

## 13. Conclusiones y decisión de inversión

### ¿Invertiríamos 1 M€ en alguno de los activos?

Si hubiera que seleccionar un único activo de los cinco analizados, el activo que presenta el perfil histórico más convincente es SPY.

Durante el periodo 2010–2026, SPY obtiene un CAGR del 14,12%, claramente superior al resto de activos, con una volatilidad anualizada del 17,05%. Además, presenta el mejor Sharpe ratio (0,86) y el mejor Calmar ratio (0,419) de la muestra.

También resulta relevante que su máximo drawdown, -33,72%, sea el menos profundo de los cinco activos estudiados, pese a ser el que consigue la mayor rentabilidad compuesta.

| Activo | CAGR | Volatilidad | Max Drawdown | Sharpe | Calmar | CRR |
|---|---:|---:|---:|---:|---:|---:|
| SPY | 14,12% | 17,05% | -33,72% | 0,860 | 0,419 | 0,514 |
| VNQ | 8,52% | 20,27% | -42,40% | 0,505 | 0,201 | 0,139 |
| GLD | 7,68% | 16,79% | -45,56% | 0,525 | 0,169 | 0,072 |
| EEM | 4,91% | 21,65% | -39,82% | 0,330 | 0,123 | 0,060 |
| TLT | 2,16% | 14,90% | -48,35% | 0,218 | 0,045 | 0,031 |

Por tanto, si la decisión consistiera obligatoriamente en escoger uno de estos cinco activos de forma individual, SPY sería el candidato más convincente según la evidencia histórica utilizada.

Esto no implica que invertir los 1 M€ íntegramente en SPY sea necesariamente la solución óptima de construcción de cartera. En una fase posterior sería razonable estudiar combinaciones entre activos, correlaciones y diversificación. La conclusión aquí se limita a comparar individualmente los cinco activos disponibles.

---

### ¿Qué activo tiene el peor perfil de cola?

La respuesta depende de qué característica de la cola queramos medir.

Si utilizamos como criterio principal la severidad histórica de las pérdidas extremas, EEM presenta el peor resultado:

- VaR histórico 95%: **2,083%**
- CVaR histórico 95%: **3,146%**

Esto significa que, dentro del peor 5% de sesiones históricas, EEM sufrió una pérdida diaria media de aproximadamente 3,15%.

Sin embargo, VNQ presenta una estructura de distribución especialmente preocupante:

- Skewness: **-0,735**
- Kurtosis: **15,645**

VNQ tiene la skewness más negativa y la kurtosis más elevada de todos los activos. Esto indica una distribución fuertemente asimétrica hacia pérdidas y con una frecuencia de observaciones extremas muy superior a la que cabría esperar bajo una distribución normal.

Por tanto, puede distinguirse entre:

- **EEM**: peor severidad histórica de cola medida mediante CVaR.
- **VNQ**: distribución con mayor asimetría negativa y colas más extremas.

---

### ¿Por qué el riesgo de cola no se aprecia mirando únicamente la volatilidad?

La volatilidad únicamente mide la dispersión global de los retornos alrededor de su media. No distingue adecuadamente entre movimientos positivos y negativos ni describe la forma de las colas.

Esto puede observarse comparando SPY y GLD:

- SPY: volatilidad **17,05%**
- GLD: volatilidad **16,79%**

Sus volatilidades son prácticamente iguales.

Sin embargo:

- Max Drawdown SPY: **-33,72%**
- Max Drawdown GLD: **-45,56%**

La similitud en volatilidad no implica, por tanto, un comportamiento similar durante episodios adversos.

El caso de VNQ es todavía más ilustrativo. Su volatilidad del 20,27% no es la mayor de la muestra, pero presenta una kurtosis de 15,65 y una skewness de -0,735, revelando una distribución con fuertes pérdidas extremas que la volatilidad por sí sola no captura.

Esto justifica complementar la volatilidad con métricas como VaR, CVaR, skewness, kurtosis y drawdown.

---

### ¿Coincide el ranking por Sharpe con el ranking por Calmar?

Los rankings son similares, pero **no coinciden completamente**.

| Activo | Sharpe Rank | Calmar Rank |
|---|---:|---:|
| SPY | 1 | 1 |
| GLD | 2 | 3 |
| VNQ | 3 | 2 |
| EEM | 4 | 4 |
| TLT | 5 | 5 |

SPY ocupa la primera posición con ambas métricas, mientras EEM y TLT mantienen las posiciones cuarta y quinta.

La diferencia aparece entre GLD y VNQ.

Por Sharpe:

1. SPY
2. GLD
3. VNQ
4. EEM
5. TLT

Por Calmar:

1. SPY
2. VNQ
3. GLD
4. EEM
5. TLT

La razón es que ambas métricas utilizan definiciones distintas de riesgo.

El Sharpe penaliza la volatilidad total. GLD se beneficia de tener una volatilidad relativamente baja, del 16,79%.

El Calmar utiliza el máximo drawdown como medida de riesgo y compara este con el crecimiento compuesto. Aunque VNQ presenta mayor volatilidad, consigue un CAGR del 8,52%, frente al 7,68% de GLD, mientras que ambos presentan drawdowns profundos.

Esto demuestra que la elección de la medida de riesgo puede alterar la evaluación relativa de los activos.

---

### ¿Cómo se comporta el VaR paramétrico en dos periodos diferentes?

El VaR gaussiano aumenta considerablemente para todos los activos al pasar del periodo 2017–2019 al periodo 2020–2022.

| Activo | VaR 2017–2019 | VaR 2020–2022 | Cambio relativo |
|---|---:|---:|---:|
| EEM | 1,68% | 2,70% | +60,74% |
| GLD | 1,05% | 1,66% | +58,84% |
| SPY | 1,27% | 2,55% | +100,75% |
| TLT | 1,05% | 1,99% | +90,14% |
| VNQ | 1,31% | 3,03% | +130,99% |

El cambio más importante se produce en VNQ, cuyo VaR aumenta aproximadamente un 131%.

SPY prácticamente duplica su VaR, mientras que TLT experimenta un incremento cercano al 90%.

Estos resultados muestran que el VaR no es una característica permanente del activo. Depende fuertemente del régimen de mercado y de las propiedades de la distribución de retornos durante el periodo utilizado para estimarlo.

Un modelo de riesgo calibrado exclusivamente utilizando 2017–2019 habría infravalorado de forma importante el riesgo observado durante 2020–2022.

Esto pone de manifiesto uno de los problemas principales de utilizar modelos de riesgo estáticos: los mercados presentan **cambios de régimen** y la volatilidad no permanece constante.

---

### ¿Qué aporta el Capital Resilience Ratio?

La métrica propuesta, Capital Resilience Ratio (CRR), combina cuatro elementos:

- crecimiento compuesto mediante CAGR,
- profundidad de las pérdidas mediante Max Drawdown,
- severidad de las pérdidas extremas mediante CVaR,
- persistencia del drawdown mediante Time Under Water.

Los resultados obtenidos son:

| Activo | CAGR | Max Drawdown | CVaR 95% | Máx. Time Under Water | CRR |
|---|---:|---:|---:|---:|---:|
| SPY | 14,12% | -33,72% | 2,59% | 488 | 0,514 |
| VNQ | 8,51% | -42,40% | 3,00% | 1.114 | 0,139 |
| GLD | 7,68% | -45,56% | 2,51% | 2.247 | 0,072 |
| EEM | 4,91% | -39,82% | 3,15% | 1.580 | 0,060 |
| TLT | 2,15% | -48,35% | 2,02% | 1.544 | 0,031 |

SPY obtiene un CRR de 0,514, muy superior al resto.

No solo genera el mayor CAGR, sino que presenta el menor máximo drawdown y recupera sus máximos históricos considerablemente antes que otros activos.

GLD ofrece un ejemplo interesante de la información adicional que aporta el CRR. Aunque ocupa la segunda posición por Sharpe, permanece hasta 2.247 sesiones consecutivas bajo máximos anteriores en el peor episodio de la muestra. Esta elevada persistencia del drawdown reduce considerablemente su CRR.

El CRR permite por tanto introducir una dimensión que métricas como Sharpe o incluso Calmar no capturan completamente:

> **el tiempo necesario para recuperar el capital perdido.**

Desde el punto de vista de un inversor, la profundidad de una pérdida no es la única dimensión relevante. También importa cuánto tiempo permanece atrapado el capital antes de volver a alcanzar un máximo.

---

## Conclusión general

El análisis muestra que no existe una única dimensión del riesgo capaz de describir completamente un activo.

La volatilidad mide dispersión, pero no caracteriza adecuadamente pérdidas extremas. El VaR identifica el comienzo de la cola, mientras que el CVaR permite estudiar su severidad. Skewness y kurtosis muestran características de la distribución que un modelo gaussiano puede ignorar. El drawdown incorpora la experiencia real de pérdida de capital, y el tiempo bajo agua añade la dimensión temporal de su recuperación.

Dentro del universo estudiado y durante el periodo 2010–2026, SPY presenta la combinación histórica más favorable entre crecimiento y riesgo, ocupando la primera posición por Sharpe, Calmar y Capital Resilience Ratio.

Al mismo tiempo, el fuerte aumento del VaR entre los periodos 2017–2019 y 2020–2022 demuestra que estas métricas deben interpretarse como estimaciones dependientes del régimen de mercado y no como parámetros permanentes de los activos.

La principal conclusión de la práctica es, por tanto, que evaluar una cartera requiere estudiar conjuntamente rentabilidad, dispersión, drawdowns, riesgo de cola y persistencia de las pérdidas, en lugar de depender de una única métrica.
